```python
from inference_sdk import InferenceHTTPClient

client = InferenceHTTPClient(
    api_url="https://serverless.roboflow.com",
    api_key=Ropoflow_api_key
)

result = client.run_workflow(
    workspace_name="detection-25dtb",
    workflow_id="custom-workflow-6",
    images={
        "image": "pre222.jpg"
    },
    use_cache=True
)
result
```

#Classes Collection

```python
result = result[0]
predictions = result['predictions']['predictions']

sorted_predictions = sorted(predictions, key=lambda x: x['x']) #sort classes with respect to x

result_text = ''.join([item['class'] for item in sorted_predictions]) #collect classes in text

print("Treatment is :", result_text)
```

#Prepare Dataset

```python
import json
with open('drugsIndex.json') as file:
    drugs = json.load(file)   #list of drugs
drugs
```

```python
names=[item['root'] for item in drugs]
names=set(names)
names
```

#clean Drug names

```python
import re
def process(drug):
  drug=re.sub(r"\W",'',drug)
  drug=re.sub(r"\d",'',drug)
  drug=re.sub(r"_",'',drug)
  drug=re.sub(r'\b\w{1,2}\b', '', drug)
  return drug
drugs=[process(drug) for drug in names]
drugs_list=[]
for item in drugs:
  if(item ==''):
    continue
  drugs_list.append(item)
```

```python
for item in drugs:
  if item == 'prosense':
    print(item)
```

#Fuzzy Matching

```python
!pip install rapidfuzz
```

```python
from rapidfuzz import process
best_match = process.extractOne(result_text, drugs_list)
print(best_match)  # ("Panadol", 92.5,index)
```

```python
if best_match[1] > 80:
    print(f"Corrected: {best_match[0]}")
else:
    print("No good match found.")
```
